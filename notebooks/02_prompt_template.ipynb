````python
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq

load_dotenv()
````

````python
llm= ChatGroq(
     model="openai/gpt-oss-120b",
    temperature=0.7,
)


print(llm)


````

##### Notes: Model Metadata and Configuration

- Meta:  `print(llm)` displays the model's configuration and metadata, such as the model name, temperature, and other settings.
- Content:  `ChatGroq()` initializes the Groq chat model with the selected model (`openai/gpt-oss-120b`) and temperature (`0.7`).

Key Metadata describes the model's configuration, while the content or response contains the actual generated text.

### Creating a `PromptTemplate`
I am creating a reusable prompt template in LangChain with dynamic inputs (`topic`, `language`, and `points`) so I can generate prompts with different values at runtime.

````python
# Define input variable names
input_variables = ["topic", "language", "points"]

# Create prompt text with placeholders
template = "Explain {topic} in {points} short points in {language}."

# Build the PromptTemplate instance
prompt = PromptTemplate(
    template=template,
    input_variables=input_variables,
)
````

##### Formatting the Prompt with `.format()`
* What it does: Uses Python's built-in `str.format` to put dynamic values into the `{...}` placeholders inside `PromptTemplate`.  

* Variables role: The values you pass (`topic`, `language`, `points`) fill in the prompt text at runtime.  

* Output: Returns a plain text string (`str`) that you can print, check, or use right away.

````python
# ২. ভেরিয়েবল বসিয়ে চূড়ান্ত প্রম্পট তৈরি
formatted_prompt = prompt.format(topic="Git", language="English", points=2)
print(formatted_prompt)
````

##### Invoking the Prompt with `.invoke()`
* What it does: Takes a dictionary of values and fills the template placeholders.  

* Output: Returns a `PromptValue` object, and `.text` gives the final prompt string.  

* Why we use it: Right now, we use `.invoke()` because it is LangChain's standard method for running a prompt and returns a `PromptValue` object, while letting us pass all inputs cleanly inside a dictionary.

````python
# ২. invoke() কল করা (ডিকশনারির ভেতরে ভেরিয়েবল দিতে হয়)
prompt_value = prompt.invoke({"topic": "Git", "language": "English", "points": 6})

# PromptValue থেকে টেক্সট বের করা
print(prompt_value.text)
````

##### Checking Output Types
* What it checks: Prints the Python data types of both `formatted_prompt` and `prompt_value`. 


* `.format()` output: Returns a standard string (`<class 'str'>`).   


* `.invoke()` output: Returns a LangChain prompt value object (`<class 'langchain_core.prompt_values.StringPromptValue'>`).

````python
print(f"Type of formatted_prompt: {type(formatted_prompt)}")
print(f"Type of prompt_value: {type(prompt_value)}")
````

##### Answer from LLM

````python
response = llm.invoke(formatted_prompt)
print(response.content)
````

### Running a Prompt + LLM Chain ( Workflow) 

```text
       Input Variables
   (topic, language, points)
              
              ▼
       Prompt Template
      ((builds a clear question))
              
              ▼
             LLM
      (generates the reply)
              
              ▼
       response.content
        (final answer)

````python
# Create a chain by connecting the prompt and LLM
# LCEL Chain: Prompt -> LLM -> Answer
chain = prompt | llm

#  Invoke the chain to get the AI response
response = chain.invoke(
    {
        "topic": "football formation",
        "language": "english",
        "points": 3
    })

# Print the generated response
print(response.content)
````

##### Notes: Running a Prompt + LLM Chain

- Pipe Operator (`|`): Connects the Prompt Template with the Language Model (LLM) in LangChain.

- Input Variables: Pass dynamic values such as `topic`, `language`, and `points` to the prompt.

- Prompt Template: Takes the input variables and formats them into a structured prompt.

- LLM: Receives the formatted prompt, processes it, and generates a response.

- `response.content`: Extracts and returns the generated text from the model's response.

Key Takeaway: The `|` operator creates a simple chain where the Prompt Template prepares the input and the LLM generates the final response.

### Difference: With and Without StrOutputParser

````python
# 1. Create an output parser
parser = StrOutputParser()

# 2. Create a chain by connecting the prompt, LLM, and parser
chain = prompt | llm | parser

# 3. Invoke the chain to get the AI response
response = chain.invoke(
    {
        "topic": "English grammar",
        "language": "english",
        "points": 6
    }
)

# 4. Print the generated response
print(response)
````



- Without `StrOutputParser`: The LLM returns an AI Message object. To access the generated text, you need to use `response.content`.

- With `StrOutputParser`: The parser extracts the generated text and returns it directly as a plain Python string.

Key `StrOutputParser` makes the output easier to use by removing the need to access `.content` manually.

#### Partial Variables 

Partial Variables allow us to pre-fill some variables in a PromptTemplate so that we don't need to provide them every time we invoke the prompt.

````python
# Set partial variables
partial_prompt = prompt.partial(
    language="English",
    points=3
)

# rovide only the remaining variable
formatted_prompt = partial_prompt.invoke({
    "topic": "Football Formation"
})

# 4. Display the formatted prompt
print(formatted_prompt.text)
````

*Why Use Partial Variables?*
- Avoid repeating fixed values.
- Make prompts shorter and cleaner.
- Keep frequently used values predefined.
- Only provide the remaining dynamic variables.

*Syntax:*
```python
partial_prompt = prompt.partial(
    language="English",
    points=3
)
```

Function Explanation
- prompt.partial(): Creates a partial prompt by assigning fixed values to selected variables.
- language="English": Sets the language variable to English.
- points=3: Sets the points variable to 3.

*Example:*

Here, `language` and `points` are fixed, while `topic` is provided dynamically.

Description: Invokes the partial_prompt by providing the remaining dynamic variable (topic). The predefined values language and points are automatically applied.

````python
formatted_prompt = partial_prompt.invoke({
    "topic": "Football Tactics"
})

print(formatted_prompt.text)
````

````python
partial_chain = partial_prompt | llm | StrOutputParser()
response = partial_chain.invoke(formatted_prompt)

# 4. Print the generated response
print(response)
````

#### Batch Processing

````python


# 1. Create a chain
batch_chain = prompt | llm | StrOutputParser()

# 2. Prepare multiple inputs
batch_inputs = [
    {
        "topic": "Football Formation",
        "language": "English",
        "points": 3
    },
    {
        "topic": "Football Tactics",
        "language": "Bangla",
        "points": 4
    },
    {
        "topic": "Offside Rule",
        "language": "English",
        "points": 5
    }
]

# 3. Process all inputs together
responses = batch_chain.batch(batch_inputs)

# 4. Display the results
for item, response in zip(batch_inputs, responses):
    print(f"\nTopic: {item['topic']}")
    print(response)
    print("-" * 50)
````
