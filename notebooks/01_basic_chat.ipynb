# 1. Basic Chat Model - LangChain

In this Jupyter Notebook, we will learn how to connect to Groq's chat model using LangChain. We will also learn how to easily send text prompts and receive responses using the `.invoke()` method.

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

#### Now, we will initialize the chat model and send a prompt or question to the model.

```python
response = llm.invoke("Do you know the capital of Bangladesh?")

print(response.content)
```

#### Let's try getting a longer and more detailed response from the model.

```python
response = llm.invoke("What mathematical concepts do I need to learn AI?")

print(response.content)
```

```python
response = llm.invoke("What is the difference between MCP and API?")

print(response.content)
```

### Code Breakdown: LLM Basic Chat Setup

This code loads Environment Variables and uses LangChain to set up and use Groq's Language Model (LLM).

---

### 1. Library Imports (Imports)

- **`from dotenv import load_dotenv`**: Imports the `load_dotenv` function from the `python-dotenv` library. It is used to read data, such as API keys, from the project's `.env` file.  

- **`from langchain_groq import ChatGroq`**: Imports the `ChatGroq` class from the `langchain-groq` package. It connects the Groq API with the LangChain framework.

---

### 2. Loading Environment Variables

- **`load_dotenv()`**: When this function is called, it loads the key-value pairs from the project's `.env` file into the system environment variables. This mainly includes the `GROQ_API_KEY`.  

  This means you do not need to write your secret API key directly inside your code.

---

### 3. Model Initialization (`ChatGroq`)

- **`llm = ChatGroq(...)`**: Creates an instance of the `ChatGroq` class named `llm`. This instance will be used later to send prompts and receive responses.  


  - **`model="openai/gpt-oss-120b"`**: Specifies which model hosted on Groq's server you want to use.  

  - **`temperature=0`**: Controls the creativity and randomness of the model's responses. Setting it to `0` helps the model produce more consistent, focused, and predictable answers with less randomness.

---

### Model Introduction: `openai/gpt-oss-120b`

**`gpt-oss-120b`** is a large open-weight language model (LLM) hosted on high-speed inference platforms like Groq.

---

### Meaning of Each Part of the Name

- **`gpt` (Generative Pre-trained Transformer):** This is a transformer-based generative model. It analyzes the input prompt and generates responses by predicting the next tokens or words.  

- **`oss` (Open-Source / Open-Weight):** This means the model's weights are publicly available. Developers can use these weights to deploy and run the model on their own hardware or inference engines, such as Groq LPU.  

- **`120b` (120 Billion Parameters):** The model contains approximately **120 billion parameters**. A higher number of parameters can help the model handle complex reasoning, coding, and in-depth analysis tasks.  


---

### Notes and Practical Usage

**Note 1 (Performance and Speed):**

Even though it is a very large model with 120 billion parameters, it can provide very fast responses because it runs on Groq's LPU (Language Processing Unit) architecture.

**Note 2 (Effect of `temperature=0`):**

Setting `temperature=0` when initializing `ChatGroq` helps the model generate consistent, focused, and information-based responses with less randomness.

**Note 3 (Suitable Use Cases):**

This model is especially useful for:

- Complex code generation.
- Logical problem-solving.
- Deep reasoning and analysis.
- Organizing data into specific formats, such as JSON or Markdown.
